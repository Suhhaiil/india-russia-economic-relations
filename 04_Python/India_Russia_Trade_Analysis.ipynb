```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    df["Financial_Year"],
    df["Total_Trade_USD_m"],
    marker="o"
)

plt.title("India–Russia Total Trade Trend (2014–2025)")
plt.xlabel("Financial Year")
plt.ylabel("Trade Value (US$ million)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df["Financial_Year"],
    df["Exports_USD_m"],
    marker="o",
    label="Exports"
)

plt.plot(
    df["Financial_Year"],
    df["Imports_USD_m"],
    marker="o",
    label="Imports"
)

plt.title("India–Russia Exports vs Imports (2014–2025)")
plt.xlabel("Financial Year")
plt.ylabel("Trade Value (US$ million)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

# -----------------------------
# 1. Key findings
# -----------------------------

summary = {
    "Average Total Trade": df["Total_Trade_USD_m"].mean(),
    "Average Exports": df["Exports_USD_m"].mean(),
    "Average Imports": df["Imports_USD_m"].mean(),
    "Average Import Share": df["Import_Share_%"].mean(),
    "Highest Total Trade": df["Total_Trade_USD_m"].max(),
    "Lowest Total Trade": df["Total_Trade_USD_m"].min(),
    "Highest Exports": df["Exports_USD_m"].max(),
    "Highest Imports": df["Imports_USD_m"].max(),
    "Largest Trade Deficit": df["Trade_Balance_USD_m"].min(),
    "Highest YoY Growth": df["YoY_Trade_Growth_%"].max(),
    "Largest YoY Decline": df["YoY_Trade_Growth_%"].min()
}

for key, value in summary.items():
    print(f"{key}: {value:.2f}")


# -----------------------------
# 2. Exports vs Imports
# -----------------------------

plt.figure(figsize=(10, 5))

plt.plot(df["Financial_Year"], df["Exports_USD_m"],
         marker="o", label="Exports")

plt.plot(df["Financial_Year"], df["Imports_USD_m"],
         marker="o", label="Imports")

plt.title("India–Russia Exports vs Imports (2014–2025)")
plt.xlabel("Financial Year")
plt.ylabel("Trade Value (US$ million)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig("exports_vs_imports.png", dpi=300)
plt.show()


# -----------------------------
# 3. Trade Balance
# -----------------------------

plt.figure(figsize=(10, 5))

plt.bar(df["Financial_Year"], df["Trade_Balance_USD_m"])

plt.title("India–Russia Trade Balance (2014–2025)")
plt.xlabel("Financial Year")
plt.ylabel("Trade Balance (US$ million)")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.tight_layout()

plt.savefig("trade_balance.png", dpi=300)
plt.show()


# -----------------------------
# 4. YoY Trade Growth
# -----------------------------

plt.figure(figsize=(10, 5))

plt.bar(df["Financial_Year"], df["YoY_Trade_Growth_%"])

plt.title("India–Russia Year-on-Year Trade Growth (2015–2025)")
plt.xlabel("Financial Year")
plt.ylabel("YoY Trade Growth (%)")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.tight_layout()

plt.savefig("yoy_trade_growth.png", dpi=300)
plt.show()


# -----------------------------
# 5. Export analysis dataset
# -----------------------------

df.to_csv("India_Russia_Trade_Python_Analysis.csv", index=False)

print("\nPython analysis complete.")
print("Files created:")
print("- exports_vs_imports.png")
print("- trade_balance.png")
print("- yoy_trade_growth.png")
print("- India_Russia_Trade_Python_Analysis.csv")
```
